```python
name = input("What is your name? ")

score = float(input("What was your test score (0-100)? "))

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print()
print("Hello,", name + "!")
print("Your test score was:", score)
print("Your letter grade is:", grade)
```
